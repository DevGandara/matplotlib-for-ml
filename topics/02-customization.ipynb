Nos enfocaremos en titulos, etiquetas, colores, estilos, leyendas, escalas y anotaciones utilizando la interfaz orientada a objetos (fig, ax)

# Setup

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
epochs = np.arange(1, 11)

train_loss = np.array([
    0.95, 0.82, 0.70, 0.58, 0.49,
    0.42, 0.36, 0.31, 0.27, 0.24
])

val_loss = np.array([
    1.00, 0.87, 0.76, 0.67, 0.61,
    0.59, 0.60, 0.63, 0.67, 0.72
])
```

# Titles y Labels
Se puede agregar un título a la grafica y etiquetas para identificar qué representa cada eje

```python
fig, ax = plt.subplots(figsize=(8,5))

ax.plot(epochs, train_loss)

ax.set_title("Training Loss", fontsize=16, fontweight="bold")
ax.set_xlabel("Epochs", fontsize=12)
ax.set_ylabel("Loss", fontsize=12)

plt.show()
```

# Colors, Line Styles y Markers

```python
fig, ax = plt.subplots(figsize=(8,5))

ax.plot(
    epochs,
    train_loss,
    color="blue",
    linewidth=0.5,
    linestyle="--",
    marker='*',
    markersize=6,
    alpha=0.4
)

plt.show()
```

# Legends

```python
fig, ax = plt.subplots(figsize=(8,5))

ax.plot(epochs, train_loss, label='Training Loss')
ax.plot(epochs, val_loss, label='Validation Loss')

ax.legend(loc='lower left')

plt.show()
```

# Grid

```python
fig, ax = plt.subplots(figsize=(8,5))

ax.plot(epochs, train_loss)

ax.grid(True, alpha=0.5, linestyle='--')

plt.show()
```

# Axis Limits y Ticks
Podemos controlar qué rango de valores se muestra y dónde aparecen las marcas de cada eje

```python
fig, ax = plt.subplots(figsize=(8,5))

ax.plot(epochs, train_loss)

ax.set_xlim(1,10)
ax.set_ylim(0,1.2)

ax.set_xticks([1, 3, 5, 7, 9])
ax.set_yticks([0, 0.25, 0.5, 0.75, 1.0])

plt.show()
```

# Axis Scales
Permite modificar la escala de los ejes
- linear: escala lineal, utilizada de manera predeterminada
- log: escala logaritmica, util para valores que cambian en varios ordenes de magnitud

```python
x = np.arange(1,100)
y = np.exp(-x/10)

fig, ax = plt.subplots(figsize=(8,5))

ax.plot(x,y)

ax.set_yscale("log")

plt.show()
```

Una escala logaritmica puede ser util para visualizar una función de pérdida que disminuye exponencialmente durante el entrenamiento. La escala logaritmica estandar requiere valores positivo

# Plot Styles
Incluye estilos predefinidos que permiten cambiar la apariencia de las gráficas sin configurar todos los paramtros manualmente


- ggplot
- dark_background
- bmh
- fivethirtyeight

```python
with plt.style.context('ggplot'):
    
    fig, ax = plt.subplots(figsize=(8,5))
    
    ax.plot(epochs, train_loss)

    ax.set_title("Training Loss")
    ax.set_xlabel("Epochs")
    ax.set_ylabel("Loss")

    plt.show()
```

# Annotations
Las anotaciones permiten resaltar puntos relevantes de una visualización. 
Por ejemplo, podemos identificar la época donde el modelo obtuvo su menor pérdida de validación

```python
best_epoch = np.argmin(val_loss)

fig, ax = plt.subplots(figsize=(8,5))

ax.plot(epochs, val_loss, marker='o')

ax.annotate(
    "Best Validation Loss",
    xy=(epochs[best_epoch], val_loss[best_epoch]),
    xytext=(7, 0.85),
    arrowprops={'arrowstyle': '->'}
)

plt.show()
```

# Ejemplos completo - ML Training Visualization

```python
import matplotlib.pyplot as plt
import numpy as np

epochs = np.arange(1,11)

train_loss = np.array([
    0.95, 0.82, 0.70, 0.58, 0.49,
    0.42, 0.36, 0.31, 0.27, 0.24
])

val_loss = np.array([
    1.00, 0.87, 0.76, 0.67, 0.61,
    0.59, 0.60, 0.63, 0.67, 0.72
])

fig, ax = plt.subplots(figsize=(9,5))

# Lines
ax.plot(
    epochs, train_loss,
    label="Training Loss",
    color="royalblue",
    linewidth=2,
    marker="o"
)

ax.plot(
    epochs, val_loss,
    label="Validation Loss",
    color="tomato",
    linewidth=2,
    linestyle="--",
    marker="s"
)

# Titles and labels
ax.set_title("Model Training History", fontsize=16)
ax.set_xlabel("Epochs")
ax.set_ylabel("Loss")

# Axis configuration
ax.set_xticks(epochs)
ax.set_ylim(0,1.1)

# Grid and legend
ax.grid(True, alpha=0.3)
ax.legend()

# Annotation
best_idx = np.argmin(val_loss)

ax.annotate(
    "Best Epoch",
    xy=(epochs[best_idx], val_loss[best_idx]),
    xytext=(1,0.9),
    arrowprops={"arrowstyle": "->"}
)

# Adjust layout
fig.tight_layout()

plt.show()
```
