```python
import matplotlib.pyplot as plt
import numpy as np

rng = np.random.default_rng(42) # Generador de numeros aleatorios reproducibles
```

# Histograms
Muestra como se distribuyen los valores numéricos mediante intervalos llamados ´bins´

```python
ages = rng.normal(
    loc=30,
    scale=8,
    size=1000
)

fig, ax = plt.subplots(figsize=(8,5))

ax.hist(ages, bins=20, edgecolor="black") # Divide datos en 20 intervalos
# Tambien puedes implementar "auto" en bins para que numpy lo haga
ax.set_title("Age Distribution")
ax.set_xlabel("Age")
ax.set_ylabel("Frequency")

plt.show()
```

## Normalized histogram
De forma predeterminada, el histograma muestra frecuencias. Podemos represetar densidad de probabilidad con density=True

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(
    ages,
    bins=25,
    density=True,
    alpha=0.7
)

plt.show()
```

# Boxplots
Un boxplot resume la distribución de los datos mediante sus cuartiles, mediana y posibles valores atipicos

```python
salaries = np.array([
    20, 22, 23, 25, 26,
    28, 29, 30, 32, 35,
    36, 38, 40, 42, 100
])

fig, ax = plt.subplots(figsize=(8, 5))

ax.boxplot(salaries)

ax.set_title("Salary Distribution")
ax.set_ylabel("Salary (thousands)")

plt.show()
```

## Comparing Distributions
Podemos comparar distintas distribuciones mediantes varios boxplots dentro de una misma figura

```python
group_a = rng.normal(70, 10, 100)
group_b = rng.normal(80, 12, 100)
group_c = rng.normal(65, 8, 100)

fig, ax = plt.subplots(figsize=(8, 5))

ax.boxplot(
    [group_a, group_b, group_c],
    tick_labels=["Group A", "Group B", "Group C"]
)

ax.set_title("Score Distribution by Group")
ax.set_ylabel("Score")

plt.show()
```

# Bar Charts
Los graficos de barras permiten comparar valores asociados a distintas categorias

```python
classes = ["Cat", "Dog", "Bird"]
counts = [120, 250, 80]

fig, ax = plt.subplots(figsize=(8, 5))

ax.bar(
    classes,
    counts,
    color=["steelblue", "orange", "green"]
)

ax.set_title("Class Distribution")
ax.set_xlabel("Class")
ax.set_ylabel("Number of Samples")

plt.show()
```

# Violin Plots
Permite visualizar la distribución de los datos y su densidad estimada.
A diferencia de un boxplot, ayuda a identificar si los datos presentan varias concentraciones o picos.

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.violinplot(
    [group_a, group_b, group_c],
    showmedians=True
)

ax.set_xticks([1, 2, 3])
ax.set_xticklabels(["Group A", "Group B", "Group C"])

ax.set_title("Distribution Comparison")
ax.set_ylabel("Score")

plt.show()
```

# Cumulative Distributions
Una ECDF muestra la proporción de observaciones menores o iguales a cada valor.
Sirve para entender percentiles y comparar distribuciones sin depender de los bins de un histograma

```python
values = np.sort(ages)
probabilities = np.arange(1, len(values) + 1) / len(values)

fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(values, probabilities)

ax.set_title("Empirical Cumulative Distribution")
ax.set_xlabel("Age")
ax.set_ylabel("Cumulative Proportion")

ax.grid(alpha=0.3)

plt.show()
```

# Complete example

```python
rng = np.random.default_rng(42)

ages = rng.normal(35, 10, 500)
income = rng.lognormal(10, 0.6, 500)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# Age distribution
axes[0, 0].hist(ages, bins=25, edgecolor="black")
axes[0, 0].set_title("Age Distribution")
axes[0, 0].set_xlabel("Age")

# Income distribution
axes[0, 1].hist(income, bins=25, edgecolor="black")
axes[0, 1].set_title("Income Distribution")
axes[0, 1].set_xlabel("Income")

# Age outliers
axes[1, 0].boxplot(ages)
axes[1, 0].set_title("Age Boxplot")

# Income outliers
axes[1, 1].boxplot(income)
axes[1, 1].set_title("Income Boxplot")

fig.tight_layout()
plt.show()
```
