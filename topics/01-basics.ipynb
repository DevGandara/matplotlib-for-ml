## ¿Qué es Matplotlib?
Es una biblioteca de Python que permite crear visualizaciones de datos mediante graficos
- Visualizar datasets antes de entrenar modelos
- Identificar relaciones entre variables
- Representar predicciones de modelos
- Analizar errores y resultados del entrenamiento

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Figure y Axes
- FIGURE: Contenedor completo de la visualización
- AXES: Área donde se dibujan los datos
- AXIS: Ejes X e Y que determinan las escalas

```python
fig, ax = plt.subplots(figsize=(8,5)) 
#fig representa la figura completa y ax graficos
# plt.plot() puede sustituir fig,ax= plt.subplots()

plt.show()
```

## Line Plot
Un grafico de líneas permite representar la relación entre variables, especialmente cuando queremos observar tendencias o cambios.
Se usa para visualizar la evolución de loss, accuracy y otras métricas durante el entrenamiento

```python
epochs = np.array([1,2,3,4,5,6])
loss = np.array([0.9,0.7,0.5,0.4,0.3,0.2])

fig, ax = plt.subplots()

ax.plot(epochs, loss) # (x,y)

plt.show()
```

## Scatter Plot
Un grafico de dispersión representa cada observación como punto.
Es particularmente importante en ML porque permite explorar relaciones entre variables numericas

```python
hours = np.array([1, 2, 3, 4, 5, 6])
scores = np.array([50, 55, 65, 70, 80, 85])

fig, ax = plt.subplots()

ax.scatter(hours, scores)

plt.show()
```

Es util para observar:
- Correlaciones o tendencias
- Agrupaciones de observaciones
- Posibles valores atipicos
- Relaciones entre caracteristicas

## Agregar varios graficos al mismo Axes

```python
x = np.array([1, 2, 3, 4, 5])
y_real = np.array([3, 5, 6, 9, 11])

# Predicción simulada
y_pred = 2 * x + 1

fig, ax = plt.subplots()

ax.scatter(x, y_real, label="Real data")
ax.plot(x, y_pred, label="Prediction", color="green")

ax.legend()

plt.show()
```
